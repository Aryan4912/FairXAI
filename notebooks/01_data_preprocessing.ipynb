```python
# ==========================================
# Feature Groups for Preprocessing
# ==========================================

target_column = "default payment next month"

# Identifier — not used as a model feature
id_column = "ID"

# Categorical demographic features
categorical_columns = [
    "SEX",
    "EDUCATION",
    "MARRIAGE"
]

# Ordinal repayment-status features
ordinal_columns = [
    "PAY_0",
    "PAY_2",
    "PAY_3",
    "PAY_4",
    "PAY_5",
    "PAY_6"
]

# Continuous numerical features
numerical_columns = [
    "LIMIT_BAL",
    "AGE",
    "BILL_AMT1",
    "BILL_AMT2",
    "BILL_AMT3",
    "BILL_AMT4",
    "BILL_AMT5",
    "BILL_AMT6",
    "PAY_AMT1",
    "PAY_AMT2",
    "PAY_AMT3",
    "PAY_AMT4",
    "PAY_AMT5",
    "PAY_AMT6"
]

# All model features
feature_columns = categorical_columns + ordinal_columns + numerical_columns

print("Feature groups defined successfully.\n")

print("ID column:")
print(id_column)

print("\nCategorical columns:")
print(categorical_columns)

print("\nOrdinal columns:")
print(ordinal_columns)

print("\nNumerical columns:")
print(numerical_columns)

print("\nTotal model features:", len(feature_columns))
print("Target column:", target_column)
```

```python
# ==========================================
# Separate Features and Target
# ==========================================

# Features
X = df[feature_columns].copy()

# Target
y = df[target_column].copy()

print("Features (X):")
print("Shape:", X.shape)

print("\nTarget (y):")
print("Shape:", y.shape)

print("\nTarget values:")
print(y.value_counts().sort_index())

print("\nFirst 5 rows of X:")
display(X.head())

print("\nFirst 5 values of y:")
display(y.head())
```

```python
# ==========================================
# Categorical Feature Categories
# ==========================================

for column in categorical_columns:
    print(f"\n{column}")
    print("Categories:", sorted(X[column].unique()))
    print("Number of categories:", X[column].nunique())
```

```python
# ==========================================
# Preprocessing Pipeline
# ==========================================

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline

# Categorical preprocessing
categorical_transformer = Pipeline(
    steps=[
        ("onehot", OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        ))
    ]
)

# Numerical + ordinal preprocessing
numerical_ordinal_columns = ordinal_columns + numerical_columns

numerical_ordinal_transformer = Pipeline(
    steps=[
        ("scaler", StandardScaler())
    ]
)

# Combine both transformations
preprocessor = ColumnTransformer(
    transformers=[
        ("categorical", categorical_transformer, categorical_columns),
        ("numerical_ordinal", numerical_ordinal_transformer, numerical_ordinal_columns)
    ],
    remainder="drop"
)

print("Preprocessing pipeline created successfully.")
print("\nCategorical columns:", categorical_columns)
print("Ordinal + numerical columns:", len(numerical_ordinal_columns))
print("Total input features:", len(feature_columns))
```

```python
# ==========================================
# Corrected Preprocessing Pipeline
# ==========================================

# Continuous numerical features that need scaling
continuous_columns = [
    "LIMIT_BAL",
    "AGE",
    "BILL_AMT1",
    "BILL_AMT2",
    "BILL_AMT3",
    "BILL_AMT4",
    "BILL_AMT5",
    "BILL_AMT6",
    "PAY_AMT1",
    "PAY_AMT2",
    "PAY_AMT3",
    "PAY_AMT4",
    "PAY_AMT5",
    "PAY_AMT6"
]

# Ordinal repayment-status features
ordinal_columns = [
    "PAY_0",
    "PAY_2",
    "PAY_3",
    "PAY_4",
    "PAY_5",
    "PAY_6"
]

# Categorical demographic features
categorical_columns = [
    "SEX",
    "EDUCATION",
    "MARRIAGE"
]

# Preprocessing for categorical features
categorical_transformer = Pipeline(
    steps=[
        ("onehot", OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        ))
    ]
)

# Preprocessing for continuous numerical features
continuous_transformer = Pipeline(
    steps=[
        ("scaler", StandardScaler())
    ]
)

# Combine transformations
preprocessor = ColumnTransformer(
    transformers=[
        ("categorical", categorical_transformer, categorical_columns),
        ("continuous", continuous_transformer, continuous_columns)
    ],
    remainder="passthrough"
)

print("Corrected preprocessing pipeline created successfully.")

print("\nCategorical columns:", len(categorical_columns))
print("Ordinal columns:", len(ordinal_columns))
print("Continuous numerical columns:", len(continuous_columns))
print("Total input features:", len(feature_columns))
```

```python
# ==========================================
# Repayment Status Code Distribution
# ==========================================

for column in ordinal_columns:
    print(f"\n{column}")
    print("-" * len(column))
    
    counts = X[column].value_counts().sort_index()
    
    display(
        pd.DataFrame({
            "Count": counts,
            "Percentage": (counts / len(X) * 100).round(2)
        })
    )
```

```python
# ==========================================
# Train-Test Split
# ==========================================

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y
)

print("Train-Test split completed.\n")

print("Training data:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTesting data:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)

print("\nTarget distribution:")

print("\nTraining set:")
print(y_train.value_counts())
print((y_train.value_counts(normalize=True) * 100).round(2))

print("\nTesting set:")
print(y_test.value_counts())
print((y_test.value_counts(normalize=True) * 100).round(2))
```

```python
# ==========================================
# Save Train-Test Split
# ==========================================

# Create the split directory
split_path = project_path / "data" / "split"
split_path.mkdir(parents=True, exist_ok=True)

# Combine features and target for easier versioning and reproducibility
train_data = X_train.copy()
train_data[target_column] = y_train

test_data = X_test.copy()
test_data[target_column] = y_test

# Save the split datasets
train_data.to_csv(split_path / "train.csv", index=False)
test_data.to_csv(split_path / "test.csv", index=False)

print("Train-test split saved successfully!\n")

print("Saved files:")
for file in sorted(split_path.iterdir()):
    print(f"- {file.name} ({file.stat().st_size / 1024:.2f} KB)")

print("\nTraining file shape:", train_data.shape)
print("Testing file shape:", test_data.shape)

print("\nSplit directory:")
print(split_path)
```

```python
# ==========================================
# Verify Saved Split
# ==========================================

# Load the versioned split files back from disk
train_check = pd.read_csv(split_path / "train.csv")
test_check = pd.read_csv(split_path / "test.csv")

print("Saved split verification:\n")

print("Train shape:", train_check.shape)
print("Test shape :", test_check.shape)

# Verify that the saved files match the in-memory datasets
train_match = train_data.reset_index(drop=True).equals(
    train_check.reset_index(drop=True)
)

test_match = test_data.reset_index(drop=True).equals(
    test_check.reset_index(drop=True)
)

print("\nTrain data matches in-memory split:", train_match)
print("Test data matches in-memory split :", test_match)

# Verify target distributions once more
print("\nTrain target distribution:")
print(train_check[target_column].value_counts())

print("\nTest target distribution:")
print(test_check[target_column].value_counts())
```

```python
# ==========================================
# Fit Preprocessor on Training Data
# ==========================================

# Fit the preprocessor ONLY on training data
# and transform the training features
X_train_processed = preprocessor.fit_transform(X_train)

# Use the SAME fitted preprocessor to transform test features
X_test_processed = preprocessor.transform(X_test)

print("Preprocessing completed successfully!\n")

print("Original training shape :", X_train.shape)
print("Processed training shape:", X_train_processed.shape)

print("\nOriginal testing shape  :", X_test.shape)
print("Processed testing shape :", X_test_processed.shape)

print("\nProcessed data type:", type(X_train_processed))
```

```python
# ==========================================
# Validate Processed Data
# ==========================================

# Check for missing and infinite values
train_nan = np.isnan(X_train_processed).sum()
test_nan = np.isnan(X_test_processed).sum()

train_inf = np.isinf(X_train_processed).sum()
test_inf = np.isinf(X_test_processed).sum()

print("Missing values:")
print("X_train_processed:", train_nan)
print("X_test_processed :", test_nan)

print("\nInfinite values:")
print("X_train_processed:", train_inf)
print("X_test_processed :", test_inf)

# Convert processed arrays to DataFrames temporarily
# using the feature names generated by the preprocessor
processed_feature_names = preprocessor.get_feature_names_out()

X_train_processed_df = pd.DataFrame(
    X_train_processed,
    columns=processed_feature_names,
    index=X_train.index
)

X_test_processed_df = pd.DataFrame(
    X_test_processed,
    columns=processed_feature_names,
    index=X_test.index
)

print("\nProcessed feature count:", len(processed_feature_names))

print("\nFirst 15 processed feature names:")
print(processed_feature_names[:15])

print("\nLast 10 processed feature names:")
print(processed_feature_names[-10:])

print("\nProcessed training data preview:")
display(X_train_processed_df.head())
```

```python
# ==========================================
# Create Final Processed Datasets
# ==========================================

# Create clean DataFrames from the processed arrays
X_train_processed_df = pd.DataFrame(
    X_train_processed,
    columns=processed_feature_names
)

X_test_processed_df = pd.DataFrame(
    X_test_processed,
    columns=processed_feature_names
)

# Add the target column back to each dataset
train_processed = X_train_processed_df.copy()
train_processed[target_column] = y_train.reset_index(drop=True)

test_processed = X_test_processed_df.copy()
test_processed[target_column] = y_test.reset_index(drop=True)

print("Final processed datasets created successfully!\n")

print("Processed training shape:", train_processed.shape)
print("Processed testing shape :", test_processed.shape)

print("\nNumber of feature columns:", len(processed_feature_names))
print("Target column:", target_column)

print("\nFirst 5 rows of processed training data:")
display(train_processed.head())

print("\nTarget distribution:")
print("\nTraining:")
print(train_processed[target_column].value_counts())

print("\nTesting:")
print(test_processed[target_column].value_counts())
```

```python
# ==========================================
# Save Processed Train-Test Data
# ==========================================

# Create the processed data directory
processed_path = project_path / "data" / "processed"
processed_path.mkdir(parents=True, exist_ok=True)

# Save processed training and testing datasets
train_processed.to_csv(
    processed_path / "train_processed.csv",
    index=False
)

test_processed.to_csv(
    processed_path / "test_processed.csv",
    index=False
)

print("Processed datasets saved successfully!\n")

print("Saved files:")
for file in sorted(processed_path.iterdir()):
    print(f"- {file.name} ({file.stat().st_size / (1024 * 1024):.2f} MB)")

print("\nTraining processed shape:", train_processed.shape)
print("Testing processed shape :", test_processed.shape)

print("\nProcessed data directory:")
print(processed_path)
```

```python
# ==========================================
# Verify Saved Processed Data
# ==========================================

# Load the processed datasets back from disk
train_processed_check = pd.read_csv(
    processed_path / "train_processed.csv"
)

test_processed_check = pd.read_csv(
    processed_path / "test_processed.csv"
)

print("Processed data verification:\n")

print("Train shape:", train_processed_check.shape)
print("Test shape :", test_processed_check.shape)

# Verify exact equality with the processed DataFrames
train_processed_match = train_processed.equals(
    train_processed_check
)

test_processed_match = test_processed.equals(
    test_processed_check
)

print("\nTrain processed data matches saved file:", train_processed_match)
print("Test processed data matches saved file :", test_processed_match)

# Check for missing and infinite values after saving/loading
train_missing = train_processed_check.isna().sum().sum()
test_missing = test_processed_check.isna().sum().sum()

train_infinite = np.isinf(
    train_processed_check.select_dtypes(include=np.number)
).sum().sum()

test_infinite = np.isinf(
    test_processed_check.select_dtypes(include=np.number)
).sum().sum()

print("\nMissing values:")
print("Train:", train_missing)
print("Test :", test_missing)

print("\nInfinite values:")
print("Train:", train_infinite)
print("Test :", test_infinite)

print("\nVerification completed.")
```

```python
# ==========================================
# Check Floating-Point Differences
# ==========================================

# Convert both DataFrames to numeric NumPy arrays
original_train_values = train_processed.to_numpy()
saved_train_values = train_processed_check.to_numpy()

original_test_values = test_processed.to_numpy()
saved_test_values = test_processed_check.to_numpy()

# Calculate the maximum absolute difference
train_max_diff = np.max(
    np.abs(original_train_values - saved_train_values)
)

test_max_diff = np.max(
    np.abs(original_test_values - saved_test_values)
)

print("Maximum absolute difference after CSV round-trip:\n")

print("Training data:", train_max_diff)
print("Testing data :", test_max_diff)

# Check whether the differences are within a very small tolerance
print("\nNumerically equivalent within tolerance:")

print(
    "Training:",
    np.allclose(
        original_train_values,
        saved_train_values,
        rtol=1e-9,
        atol=1e-12
    )
)

print(
    "Testing :",
    np.allclose(
        original_test_values,
        saved_test_values,
        rtol=1e-9,
        atol=1e-12
    )
)
```

```python
# ==========================================
# Save Fitted Preprocessing Pipeline
# ==========================================

import joblib

# Create the artifacts directory
artifacts_path = project_path / "artifacts"
artifacts_path.mkdir(parents=True, exist_ok=True)

# Save the fitted preprocessor
preprocessor_path = artifacts_path / "preprocessor.pkl"

joblib.dump(preprocessor, preprocessor_path)

print("Fitted preprocessor saved successfully!\n")

print("File:", preprocessor_path)
print("Size:", round(preprocessor_path.stat().st_size / 1024, 2), "KB")

# Verify that the file exists
print("\nFile exists:", preprocessor_path.exists())
```

```python
# ==========================================
# Verify Saved Preprocessor
# ==========================================

# Load the fitted preprocessor back from disk
preprocessor_check = joblib.load(preprocessor_path)

# Transform the same training and testing data
X_train_check = preprocessor_check.transform(X_train)
X_test_check = preprocessor_check.transform(X_test)

# Compare with the transformations produced earlier
train_preprocessing_match = np.allclose(
    X_train_processed,
    X_train_check,
    rtol=1e-9,
    atol=1e-12
)

test_preprocessing_match = np.allclose(
    X_test_processed,
    X_test_check,
    rtol=1e-9,
    atol=1e-12
)

print("Saved preprocessor verification:\n")

print("Training transformation matches:", train_preprocessing_match)
print("Testing transformation matches :", test_preprocessing_match)

print("\nOriginal processed shapes:")
print("X_train_processed:", X_train_processed.shape)
print("X_test_processed :", X_test_processed.shape)

print("\nReloaded transformation shapes:")
print("X_train_check:", X_train_check.shape)
print("X_test_check :", X_test_check.shape)
```
